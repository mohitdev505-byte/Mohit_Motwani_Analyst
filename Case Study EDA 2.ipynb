```python
activity = pd.read_csv("dailyActivity_merged.csv")
calories = pd.read_csv("dailyCalories_merged.csv")
intensities = pd.read_csv("dailyIntensities_merged.csv")
steps = pd.read_csv("dailySteps_merged.csv")
sleep = pd.read_csv("sleepDay_merged.csv")
weight = pd.read_csv("weightLogInfo_merged.csv")
heartrate = pd.read_csv("heartrate_seconds_merged.csv")
```

```python
activity["ActivityDate"] = pd.to_datetime(activity["ActivityDate"])
calories["ActivityDay"] = pd.to_datetime(calories["ActivityDay"])
intensities["ActivityDay"] = pd.to_datetime(intensities["ActivityDay"])
steps["ActivityDay"] = pd.to_datetime(steps["ActivityDay"])
sleep["SleepDay"] = pd.to_datetime(sleep["SleepDay"])
weight["Date"] = pd.to_datetime(weight["Date"])
heartrate["Time"] = pd.to_datetime(heartrate["Time"])
```

```python
heartrate_daily = (
    heartrate
    .assign(Date=heartrate["Time"].dt.date)
    .groupby(["Id", "Date"])["Value"]
    .mean()
    .reset_index()
)

heartrate_daily["Date"] = pd.to_datetime(heartrate_daily["Date"])

heartrate_daily
```

```python
df = activity.merge(calories, left_on=["Id", "ActivityDate"],
                    right_on=["Id", "ActivityDay"], how="left")

df = df.merge(intensities, on=["Id", "ActivityDay"], how="left")
df = df.merge(steps, on=["Id", "ActivityDay"], how="left")

df.rename(columns={"ActivityDate": "Date"}, inplace=True)

df
```

```python
# Add specific suffixes to each merge operation to avoid duplicate column names
df = df.merge(sleep, left_on=["Id", "Date"],
              right_on=["Id", "SleepDay"], how="left", 
              suffixes=('', '_sleep'))  # Use empty string for left df and '_sleep' for right df

df = df.merge(weight, left_on=["Id", "Date"],
              right_on=["Id", "Date"], how="left",
              suffixes=('', '_weight'))  # Use empty string for left df and '_weight' for right df

df = df.merge(heartrate_daily, on=["Id", "Date"], how="left",
              suffixes=('', '_hr'))  # Use empty string for left df and '_hr' for right df

df
```
